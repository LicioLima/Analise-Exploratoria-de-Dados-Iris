```python
# importing packages

import seaborn as sns

import matplotlib.pyplot as plt

sns.scatterplot(x="SepalLengthCm", y="SepalWidthCm", hue="Species", data=df, palette='Set2')

# placing legend outside the figure
plt.legend(bbox_to_anchor=(1, 1), loc=2)
plt.show()
```

```python
import seaborn as sns

import matplotlib.pyplot as plt

sns.scatterplot(x="PetalLengthCm", y="PetalWidthCm", hue="Species", data=df, palette='Set2')

# placing legend outside the figure
plt.legend(bbox_to_anchor=(1, 1), loc=2)
plt.show()
```

```python
# Importing packages

import seaborn as sns

import matplotlib.pyplot as plt

sns.pairplot(df.drop(['Id'], axis=1), hue='Species', height = 2)

plt.show()
```

```python
# importing packages
import seaborn as sns

import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(10, 10))

axes [0,0].set_title("Sepal Length")
axes [0,0].hist(df["SepalLengthCm"], bins=7)

axes [0,0].set_title("Sepal Width")
axes [0,0].hist(df["SepalWidthCm"], bins=5)

axes [0,0].set_title("Petal Length")
axes [0,0].hist(df["PetalLengthCm"], bins=6)

axes [0,0].set_title("Petal Width")
axes [0,0].hist(df["PetalWidthCm"], bins=6)
```

```python
import seaborn as sns

import matplotlib.pyplot as plt

plot = sns.FacetGrid(df, hue="Species")

plot.map(sns.distplot, "SepalLengthCm").add_legend()


plot = sns.FacetGrid(df, hue="Species")

plot.map(sns.distplot, "SepalWidthCm").add_legend()
```

```python
# Calculando a correlação apenas para as colunas numéricas do DataFrame original
df.drop('Id', axis=1).corr(method='pearson', numeric_only=True)
```

```python
# Importing packages
import seaborn as sns
import matplotlib.pyplot as plt

# Calculando a correlação apenas para colunas numéricas e excluindo o 'Id'
corr_matrix = df.drop(['Id'], axis=1).corr(method='pearson', numeric_only=True)

sns.heatmap(corr_matrix, annot=True, cmap='RdYlGn')
plt.show()
```

```python
#importing packages
import seaborn as sns
import matplotlib.pyplot as plt

def graph(y):
    sns.boxplot(x='Species', y=y, data=df, palette='Set2')

plt.figure(figsize=(10,10))


# Adding the subplot at the specified grid position
plt.subplot(2,2,1)
graph('SepalLengthCm')

plt.subplot(2,2,2)
graph('SepalWidthCm')

plt.subplot(2,2,3)
graph('PetalLengthCm')

plt.subplot(2,2,4)
graph('PetalWidthCm')

plt.show()
```
